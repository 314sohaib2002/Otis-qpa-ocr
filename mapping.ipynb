```python
# Assuming 'sheet' and 'df_admin' are still active

def clean_val(val):
    if val is None or val == "-":
        return ""
    return str(val)

total_overtime = clean_val(sheet["F36"].value)

# For G36/H2+I2 logic
raw_overtime_payout = sheet["G36"].value 
if raw_overtime_payout is None or raw_overtime_payout == "-":
    try:
        h2 = sheet["H2"].value if isinstance(sheet["H2"].value, (int, float)) else 0
        i2 = sheet["I2"].value if isinstance(sheet["I2"].value, (int, float)) else 0
        overtime_payout = str(h2 + i2)
    except:
        overtime_payout = "0"
else:
    overtime_payout = clean_val(raw_overtime_payout)

overtime_50 = clean_val(sheet["H36"].value)
overtime_100 = clean_val(sheet["I36"].value)

# Time balance transferred
time_balance = clean_val(sheet["E40"].value)

# Vacation days and remaining
days_taken = clean_val(sheet["K36"].value)
remaining_holiday = clean_val(sheet["K41"].value)

# ----------------- FIXED TIMEBANK LOGIC -----------------
# Both columns will now strictly use E5 - E40 as requested.
e5_val = sheet["E5"].value if isinstance(sheet["E5"].value, (int, float)) else 0
e40_val = sheet["E40"].value if isinstance(sheet["E40"].value, (int, float)) else 0

# (E5 - E40) calculates to 20 - 20 = 0 based on the current Excel file
timebank_math_result = str(e5_val - e40_val)

# Assign the math result to the relevant columns
timebank_payout = timebank_math_result
time_off_withdrawal = timebank_math_result

# Overtime hours for time off (E38)
overtime_for_time_off = clean_val(sheet["E38"].value)
# --------------------------------------------------------

# Advanced Parsing for Dates (Rows 4 through 34)
vacation_dates = []
self_report_dates = []
sick_leave_dates = []

for row in range(4, 35):
    day_val = sheet[f"A{row}"].value
    
    if not isinstance(day_val, (int, float)):
        continue
        
    day_int = int(day_val)
    
    # Check Vacation (Column K)
    k_val = sheet[f"K{row}"].value
    if isinstance(k_val, (int, float)) and k_val > 0:
        vacation_dates.append(str(day_int))
        
    # Check Sick Leave / Absence (Column J) and Comments (Column D)
    j_val = sheet[f"J{row}"].value
    d_val = sheet[f"D{row}"].value
    
    if isinstance(j_val, (int, float)) and j_val > 0:
        if d_val and str(d_val).strip() != "":
            sick_leave_dates.append(str(day_int))
        else:
            self_report_dates.append(str(day_int))

vacation_dates_str = ", ".join(vacation_dates)
self_report_dates_str = ", ".join(self_report_dates)
sick_leave_dates_str = ", ".join(sick_leave_dates)

# Update the Last Row in df_admin
last_idx = df_admin.index[-1]
cols = df_admin.columns

df_admin.loc[last_idx, cols[5]] = total_overtime        
df_admin.loc[last_idx, cols[6]] = overtime_payout       
df_admin.loc[last_idx, cols[7]] = overtime_50           
df_admin.loc[last_idx, cols[8]] = overtime_100          

df_admin.loc[last_idx, cols[9]] = overtime_for_time_off 
df_admin.loc[last_idx, cols[10]] = time_off_withdrawal  
df_admin.loc[last_idx, cols[11]] = timebank_payout      

df_admin.loc[last_idx, cols[12]] = time_balance         
df_admin.loc[last_idx, cols[13]] = vacation_dates_str   
df_admin.loc[last_idx, cols[14]] = days_taken           
df_admin.loc[last_idx, cols[15]] = remaining_holiday    

df_admin.loc[last_idx, cols[16]] = self_report_dates_str 
df_admin.loc[last_idx, cols[17]] = sick_leave_dates_str  

print(f"Calculated value for E5 - E40 (applied to Timebank & Withdrawal columns): {timebank_math_result}")
print("\nFully Updated Admin Template:")
display(df_admin.iloc[:, :18])
```

```python
import pandas as pd
from datetime import datetime

# Assuming 'sheet' (from openpyxl) is still active from your input_excel
employee_id = clean_val(sheet["F2"].value)
employee_name = clean_val(sheet["B2"].value)

# Let's write a small helper function to group consecutive days.
# If an employee takes days 10, 11, 12, 13, 14, this groups them into "From 10" to "To 14" (5 days)
def group_consecutive_days(days_list):
    if not days_list:
        return []
        
    days_list = sorted([int(d) for d in days_list])
    groups = []
    
    start_day = days_list[0]
    prev_day = days_list[0]
    count = 1
    
    for day in days_list[1:]:
        if day == prev_day + 1:
            # It is consecutive
            count += 1
            prev_day = day
        else:
            # Gap found, save the previous group
            groups.append({"from": start_day, "to": prev_day, "days": count})
            start_day = day
            prev_day = day
            count = 1
            
    # Append the last group
    groups.append({"from": start_day, "to": prev_day, "days": count})
    return groups


# 1. First, we need to gather the actual days from the input sheet again
vacation_days = []
self_report_sick = [] # No comment in column D
cert_sick = []        # Has comment in column D

for row in range(4, 35):
    day_val = sheet[f"A{row}"].value
    if not isinstance(day_val, (int, float)):
        continue
        
    day_int = int(day_val)
    
    # Vacation
    k_val = sheet[f"K{row}"].value
    if isinstance(k_val, (int, float)) and k_val > 0:
        vacation_days.append(day_int)
        
    # Sick
    j_val = sheet[f"J{row}"].value
    d_val = sheet[f"D{row}"].value
    if isinstance(j_val, (int, float)) and j_val > 0:
        if d_val and str(d_val).strip() != "":
            cert_sick.append(day_int)
        else:
            self_report_sick.append(day_int)

# 2. Group the days
vacation_groups = group_consecutive_days(vacation_days)
self_report_groups = group_consecutive_days(self_report_sick)
cert_sick_groups = group_consecutive_days(cert_sick)

# 3. Create rows for the Absence template
# Looking at the template, the columns line up as:
# Index 1: 'ee nr'
# Index 2: 'Employee Name'
# Index 4: 'Absence type'
# Index 5: '%'
# Index 6: 'date from'
# Index 7: 'date to'
# Index 8: 'number of days'
# Index 9: 'Internal comment'

absence_rows = []

month_str = clean_val(sheet["I2"].value) # e.g. "Jun"
year_str = clean_val(sheet["J2"].value)  # e.g. "2026"

def format_date(day):
    # Formats the simple integer day into e.g. "10-Jun-2026"
    return f"{day}-{month_str}-{year_str}"

# Add Vacation rows
for grp in vacation_groups:
    row_data = [None] * len(df_absence.columns)
    row_data[1] = employee_id
    row_data[2] = employee_name
    row_data[4] = "Ferie (Vacation)"  # Or whatever code you use for vacation
    row_data[5] = "100%"
    row_data[6] = format_date(grp["from"])
    row_data[7] = format_date(grp["to"])
    row_data[8] = grp["days"]
    absence_rows.append(row_data)

# Add Self-Report Sick rows (Code 2 based on your template text)
for grp in self_report_groups:
    row_data = [None] * len(df_absence.columns)
    row_data[1] = employee_id
    row_data[2] = employee_name
    row_data[4] = "2" # "Code 2" from template instructions
    row_data[5] = "100%"
    row_data[6] = format_date(grp["from"])
    row_data[7] = format_date(grp["to"])
    row_data[8] = grp["days"]
    absence_rows.append(row_data)

# Add Certified Sick rows
for grp in cert_sick_groups:
    row_data = [None] * len(df_absence.columns)
    row_data[1] = employee_id
    row_data[2] = employee_name
    row_data[4] = "Syk med legeerklæring" # Sickness with medical certificate
    row_data[5] = "100%"
    row_data[6] = format_date(grp["from"])
    row_data[7] = format_date(grp["to"])
    row_data[8] = grp["days"]
    absence_rows.append(row_data)

# 4. Append to df_absence
if absence_rows:
    new_df = pd.DataFrame(absence_rows, columns=df_absence.columns)
    df_absence = pd.concat([df_absence, new_df], ignore_index=True)

print("Updated Absence Template Data (Showing the new rows):")
display(df_absence.tail(max(1, len(absence_rows))))
```
